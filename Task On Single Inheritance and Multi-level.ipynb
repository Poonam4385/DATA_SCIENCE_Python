6. Class Name: BankAccount → SavingsAccount
Problem
Create a parent class named BankAccount.
Explanation
Provide methods:
• deposit()
• withdraw()
Create subclass:
• SavingsAccount
Add:
• calculate_interest()

```python
class BankAccount:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Balance:", self.balance)

    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            print("Balance:", self.balance)
        else:
            print("Invalid withdrawal")


class SavingsAccount(BankAccount):

    def calculate_interest(self):
        return self.balance * 0.04


account = SavingsAccount(10000)

account.deposit(2000)
account.withdraw(3000)

print("Interest:", account.calculate_interest())
```

29. Class Name:
Shape → TwoDShape → Triangle
Problem
Create multi-level inheritance.
Explanation
Triangle should calculate its area using Heron's Formula.
Store:
• side1
• side2
• side3
Implement the area calculation.

```python
import math

class Shape:

    def display(self):
        print("Shape")


class TwoDShape(Shape):

    def show_dimension(self):
        print("This is a 2D shape")


class Triangle(TwoDShape):

    def __init__(self, side1, side2, side3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3

    def area(self):

        s = (
            self.side1
            + self.side2
            + self.side3
        ) / 2

        return math.sqrt(
            s
            * (s - self.side1)
            * (s - self.side2)
            * (s - self.side3)
        )


triangle = Triangle(3, 4, 5)

print("Triangle Area:", triangle.area())
```

30. Class Name:
GrandParent → Parent → Child → GrandChild
Problem
Create four levels of inheritance.
Explanation
Each class should have its own method.
The GrandChild object should be able to access methods from all parent classes.
Demonstrate Python's Method Resolution Order (MRO) using:
ClassName.mro()
or
help(ClassName)

```python
class GrandParent:

    def grandparent_method(self):
        print("Method from GrandParent")


class Parent(GrandParent):

    def parent_method(self):
        print("Method from Parent")


class Child(Parent):

    def child_method(self):
        print("Method from Child")


class GrandChild(Child):

    def grandchild_method(self):
        print("Method from GrandChild")


obj = GrandChild()

obj.grandparent_method()
obj.parent_method()
obj.child_method()
obj.grandchild_method()


print("\nMRO:")

print(GrandChild.mro())
```
