```python
# 14. BankAccount using encapsulation

class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")

        elif amount > self.__balance:
            print("Insufficient balance.")

        else:
            self.__balance -= amount
            print("Withdrawn:", amount)

    def show_balance(self):
        print("Current balance:", self.__balance)

account = BankAccount(10000)
account.show_balance()
account.deposit(5000)
account.withdraw(3000)
account.show_balance()
```

```python
#18. Create class Rectangle with private length and width. Add getter, setter and method to calculate area.
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def set_length(self, length):
        if length > 0:
            self.__length = length
        else:
            print("Length must be positive.")

    def get_width(self):
        return self.__width

    def set_width(self, width):
        if width > 0:
            self.__width = width
        else:
            print("Width must be positive.")

    def area(self):
        return self.__length * self.__width


rect = Rectangle(10, 5)

print("Length:", rect.get_length())
print("Width:", rect.get_width())
print("Area:", rect.area())

rect.set_length(20)
rect.set_width(10)

print("New Area:", rect.area())
```

```python
#20. Create multi-level inheritance: Person → Employee → Manager.
class Person:
    def __init__(self, name):
        self.name = name


class Employee(Person):
    def __init__(self, name, employee_id):
        super().__init__(name)
        self.employee_id = employee_id


class Manager(Employee):
    def __init__(self, name, employee_id, department):
        super().__init__(name, employee_id)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Department:", self.department)


manager = Manager("Amit", 101, "IT")

manager.display()
```

```python
#23. Create abstract class Shape with abstract method area(). Implement it in Circle and Rectangle using super().
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__("Rectangle")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


circle = Circle(5)
rectangle = Rectangle(10, 5)

print(circle.name, "Area =", circle.area())
print(rectangle.name, "Area =", rectangle.area())
```

```python
#27. Create class Point and overload + operator to add two points.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print("(", self.x, ",", self.y, ")")


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

print("First point:")
p1.display()

print("Second point:")
p2.display()

print("After addition:")
p3.display()
```

```python
#28. Create class Book and overload > operator to compare number of pages of two books.
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __gt__(self, other):
        return self.pages > other.pages


book1 = Book("Python Basics", 300)
book2 = Book("Advanced Python", 450)

if book1 > book2:
    print(book1.title, "has more pages.")
else:
    print(book2.title, "has more pages.")
```

```python
#30. Create class ComplexNumber and overload both + and - operators.
class ComplexNumber:
    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imaginary + other.imaginary
        )

    def __sub__(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imaginary - other.imaginary
        )

    def display(self):
        if self.imaginary >= 0:
            print(f"{self.real} + {self.imaginary}i")
        else:
            print(f"{self.real} - {abs(self.imaginary)}i")


c1 = ComplexNumber(5, 3)
c2 = ComplexNumber(2, 1)

addition = c1 + c2
subtraction = c1 - c2

print("First complex number:")
c1.display()

print("Second complex number:")
c2.display()

print("Addition:")
addition.display()

print("Subtraction:")
subtraction.display()
```
