```python
# take the url of the page
url ="https://www.amazon.in/s?k=laptops&crid=ARB5NM49P398&sprefix=%2Caps%2C556&ref=nb_sb_ss_recent_1_0_recent"
```

```python
# creat the request header
headers={"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36"}
```

```python
# Send the request
response = requests.get(url,headers=headers)

# Check if the request was successful
response.status_code
```

```python
# Beautifulsoup object
from bs4 import BeautifulSoup
bs = BeautifulSoup(response.content,"html.parser")
```

```python
# Create the empty list to store the data
Data = []
```

```python
# Parse the content of the page
for page in range (1,3):
    
# Parameter for the page number
 params = {"k":"laptops","page":page}

# Sent the request for each page
response = requests.get(url,headers=headers,params=params)

# store product container in a variable class:class="div",class="sg-col-inner"
product_containers=bs.find_all("div",class_="sg-col-inner")

# loop through the product containers
for product in product_containers:
    title_tag=product.find("h2")
    if not title_tag:
        continue
    title= title_tag.get_text(" ",strip=True)
    if not title:
        continue

    # Step 2: Extract the price of the laptop
    price_tag = product.find("span",class_="a-price-whole")
    if price_tag:
        price= price_tag.get_text(" ",strip=True)
    else:
        price="N/A"
        
    # Step 3: Extract the rating of the laptop
    rating_tag=product.find("span",class_="a-size-small a-color-base")
    if rating_tag:
        rating=rating_tag.get_text(" ",strip=True)
    else:
        rating="N/A"
        
    # Step 4: Extract the brand name from the title using regex
    match = re.search(r"^([A-Za-z]+)",title)
    brand=match.group(1)if match else "Unknown"
    
    # Extract RAM from product title
    ram_match = re.search(r"(\d+\s*GB\s*(?:DDR\d+|LPDDR\d+)?)", title, re.IGNORECASE)
    ram=ram_match.group(1) if ram_match else "N/A"
    
   

    # Stored the data in the list
    Data.append({"title":title,
                "price":price,
                "rating":rating,
                "brand":brand,
                "ram":ram})
        
```

```python
for item in Data:
    print(f"Title:{item["title"]}")
    print(f"Price:{item["price"]}")
    print(f"Rating:{item["rating"]}")
    print(f"Brand:{item["brand"]}")
    print(f"Ram:{item["ram"]}")
    print("_" * 50)
```
