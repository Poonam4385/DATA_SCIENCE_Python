# Task on Nested Conditions (if, if-else, if-elif-else)


```python
#1. Take a number as input. If the number is positive, check whether it is even or odd and print the result.
num = int(input("Enter a number: "))

if num > 0:
    if num % 2 == 0:
        print("Positive Even Number")
    else:
        print("Positive Odd Number")
else:
    print("Number is not positive")
```

```python
#2. Take age as input. If age is greater than or equal to 18, check if the person is a senior citizen (age >= 60) or adult.
age = int(input("Enter age: "))

if age >= 18:
    if age >= 60:
        print("Senior Citizen")
    else:
        print("Adult")
else:
    print("Minor")
```

```python
# 3. Take a single character. If it is an alphabet, check whether it is uppercase or lowercase.
char = input("Enter a character: ")

if char.isalpha():
    if char.isupper():
        print("Uppercase Alphabet")
    else:
        print("Lowercase Alphabet")
else:
    print("Not an alphabet")
```

```python
# 4. Take three numbers as input and print the largest number using nested if statements (without using max()).
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b:
    if a > c:
        print("Largest number is:", a)
    else:
        print("Largest number is:", c)
else:
    if b > c:
        print("Largest number is:", b)
    else:
        print("Largest number is:", c)
```

```python
# 5. Take student marks. If marks >= 40 (Pass), then further check if marks >= 75 to print "Pass with Distinction".
marks = int(input("Enter marks: "))

if marks >= 40:
    print("Pass")
    if marks >= 75:
        print("Pass with Distinction")
else:
    print("Fail")
```

```python
# 8. Take three sides of a triangle. First check if they form a valid triangle, then check if it is Equilateral, Isosceles, or Scalene.
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    print("Valid Triangle")

    if a == b and b == c:
        print("Equilateral Triangle")
    elif a == b or b == c or a == c:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")
else:
    print("Invalid Triangle")
```

```python
# 10.Take a character. If it is a letter, check if it is vowel or consonant. If it is a digit, print "Digit". Else print "Special Character".
char = input("Enter a character: ")

if char.isalpha():
    if char.lower() in "aeiou":
        print("Vowel")
    else:
        print("Consonant")
elif char.isdigit():
    print("Digit")
else:
    print("Special Character")
```

```python
# 11.Take a year as input. Check if it is a leap year using nested conditions.
year = int(input("Enter year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap Year")
        else:
            print("Not a Leap Year")
    else:
        print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
# 13.Take three numbers. Check if all three are equal, or any two are equal using nested if.
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a == b:
    if b == c:
        print("All three are equal")
    else:
        print("Any two are equal")
else:
    if a == c or b == c:
        print("Any two are equal")
    else:
        print("All are different")
```

```python
# 18.Take three sides of a triangle. Check if valid, then check if it is a right-angled triangle using Pythagoras theorem.
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and a + c > b and b + c > a:
    print("Valid Triangle")

    if a*a + b*b == c*c or a*a + c*c == b*b or b*b + c*c == a*a:
        print("Right-angled Triangle")
    else:
        print("Not a Right-angled Triangle")
else:
    print("Invalid Triangle")
```
