```python
#22. (Hard) Create a tuple with 8 numbers. Find the index of the number25. If it exists, count how many times it appears. 
#Then convert the tupleinto a list, sort it in descending order, and print.
numbers = (10, 25, 30, 25, 40, 50, 25, 60)

if 25 in numbers:
    print("Index of 25:", numbers.index(25))
    print("Count of 25:", numbers.count(25))
else:
    print("25 does not exist in the tuple")

numbers_list = list(numbers)

numbers_list.sort(reverse=True)

print("Sorted list in descending order:", numbers_list)
```

```python
#23. (Hard) Take 4 student names and their marks as input from the user.Store them in a dictionary (name as key, marks as value). 
# Then find andprint the student with the highest marks.
students = {}

for i in range(4):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks

highest_student = max(students, key=students.get)

print("Student marks:", students)
print("Highest marks student:", highest_student)
print("Marks:", students[highest_student])
```

```python
# 24. (Hard) Create a dictionary info = {"name": "Aman", "age": 21, "city":"Mumbai"}. Make a copy of it. Clear the original dictionary. 
#Then add a new key "country": "India" in the copy and print both dictionaries.
info = {"name": "Aman","age": 21,"city": "Mumbai"}

copy_info = info.copy()

info.clear()

copy_info["country"] = "India"

print("Original dictionary:", info)
print("Copied dictionary:", copy_info)
```

```python
#25. (Hard) Given data = (10, 20, 30, 20, 40, 20, 50), convert it into a dictionary where key is the number and value is its count of occurrences (using tuple methods + dictionary)
data = (10, 20, 30, 20, 40, 20, 50)

count_dict = {}

for num in data:
    count_dict[num] = data.count(num)

print(count_dict)
```
