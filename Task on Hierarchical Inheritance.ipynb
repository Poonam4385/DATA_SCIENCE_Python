1. Person → Student, Teacher

and Teacher that inherit from Person.

16. Shape Area
Concept
Polymorphism using overriding.
How to Solve
1. Parent constructor.
2. Circle calculates πr2.
3. Square calculates side2.
4. Override area().

```python
import math

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

c = Circle(5)
s = Square(4)

print(c.area())
print(s.area())
```

General Formula to Solve Any Hierarchical Inheritance Question
For every hierarchical inheritance problem, follow this sequence:
1. Identify the Parent Class (common attributes and methods).
2. Identify the Child Classes (specialized versions of the parent).
3. Write the Parent Constructor (__init__()) if attributes are needed.
4. Use super() in child constructors to initialize parent attributes.
5. Add Child-Specific Attributes/Methods.
6. Override Methods if the child needs different behavior.
7. Create Objects of each child class.
8. Call Parent Methods to verify inheritance.
9. Call Child Methods to verify specialization.
10. Print Results to confirm the program works correctly.

```python
class Parent:
    def __init__(self, common_attribute):
        self.common_attribute = common_attribute

    def common_method(self):
        print("Common method from Parent")


class Child1(Parent):
    def __init__(self, common_attribute, child1_attribute):
        super().__init__(common_attribute)
        self.child1_attribute = child1_attribute

    def child1_method(self):
        print("Child1 specific method")


class Child2(Parent):
    def __init__(self, common_attribute, child2_attribute):
        super().__init__(common_attribute)
        self.child2_attribute = child2_attribute

    def child2_method(self):
        print("Child2 specific method")


obj1 = Child1("Common Data", "Child1 Data")
obj2 = Child2("Common Data", "Child2 Data")

# Parent method inherited by both children
obj1.common_method()
obj2.common_method()

# Child-specific methods
obj1.child1_method()
obj2.child2_method()

# Inherited and child attributes
print(obj1.common_attribute)
print(obj1.child1_attribute)

print(obj2.common_attribute)
print(obj2.child2_attribute)
```
