30. Interview Challenge
Given:
students = [
{"name":"Ram","marks":[70,80,90]},
{"name":"Amit","marks":[90,92,94]},
{"name":"Neha","marks":[60,65,70]},

Sumit Sir Notes
{"name":"Raj","marks":[88,84,90]}
]
Using only lambda with built-in functions (map, filter, sorted, max, min, reduce where
appropriate):
1. Calculate each student's average marks.
2. Sort students by average marks (highest first).
3. Find the topper.
4. Filter students whose average is greater than 80.
5. Print only the names of the filtered students.

```python
students = [
    {"name": "Ram", "marks": [70, 80, 90]},
    {"name": "Amit", "marks": [90, 92, 94]},
    {"name": "Neha", "marks": [60, 65, 70]},
    {"name": "Raj", "marks": [88, 84, 90]}
]

# Function to calculate average marks
average = lambda s: sum(s["marks"]) / len(s["marks"])


# 1. Calculate each student's average
averages = list(map(lambda s: (s["name"], round(average(s), 2)), students))

print("Averages:")
print(averages)


# 2. Sort students by average marks - highest first
ranked = sorted(
    students,
    key=lambda s: average(s),
    reverse=True
)

print("\nRanking:")
print(
    list(
        map(
            lambda s: (s["name"], round(average(s), 2)),
            ranked
        )
    )
)


# 3. Find topper
topper = max(
    students,
    key=lambda s: average(s)
)

print("\nTopper:")
print(topper["name"])


# 4. Filter students having average > 80
above_80 = list(
    filter(
        lambda s: average(s) > 80,
        students
    )
)

print("\nStudents with average > 80:")
print(above_80)


# 5. Print only names
names = list(
    map(
        lambda s: s["name"],
        above_80
    )
)

print("\nNames:")
print(names)
```

2. Employee Salary Analytics (★★★★★)
employees = [
{"id":101,"name":"Ram","dept":"IT","salary":55000},
{"id":102,"name":"Amit","dept":"HR","salary":40000},
{"id":103,"name":"Neha","dept":"IT","salary":70000},
{"id":104,"name":"Raj","dept":"Sales","salary":45000},
{"id":105,"name":"Anu","dept":"IT","salary":85000},
{"id":106,"name":"Kiran","dept":"Sales","salary":62000},

Sumit Sir Notes
]
Without using loops.
Using only lambda and built-in functions,
Find
1. Highest salary employee
2. Lowest salary employee
3. Average salary
4. Employees earning above average
5. Sort by department then salary descending
6. Total IT salary
7. Print only employee names
8. Find second highest salary

```python
from functools import reduce

employees = [
    {"id": 101, "name": "Ram", "dept": "IT", "salary": 55000},
    {"id": 102, "name": "Amit", "dept": "HR", "salary": 40000},
    {"id": 103, "name": "Neha", "dept": "IT", "salary": 70000},
    {"id": 104, "name": "Raj", "dept": "Sales", "salary": 45000},
    {"id": 105, "name": "Anu", "dept": "IT", "salary": 85000},
    {"id": 106, "name": "Kiran", "dept": "Sales", "salary": 62000}]

# 1. Highest salary employee
highest = max(employees, key=lambda e: e["salary"])
print("Highest Salary Employee:", highest)


# 2. Lowest salary employee
lowest = min(employees, key=lambda e: e["salary"])
print("Lowest Salary Employee:", lowest)


# 3. Average salary
total_salary = reduce(
    lambda total, e: total + e["salary"],
    employees,
    0)

average_salary = total_salary / len(employees)

print("Average Salary:", average_salary)


# 4. Employees earning above average
above_average = list(
    filter(
        lambda e: e["salary"] > average_salary,
        employees))

print(
    "Employees Above Average:",
    list(map(lambda e: e["name"], above_average)))


# 5. Sort by department, then salary descending
sorted_employees = sorted(
    employees,
    key=lambda e: (e["dept"], -e["salary"]))

print(
    "Sorted Employees:",
    list(
        map(
            lambda e: (e["name"], e["dept"], e["salary"]),
            sorted_employees)))


# 6. Total IT salary
it_employees = filter(
    lambda e: e["dept"] == "IT",
    employees)

total_it_salary = reduce(
    lambda total, e: total + e["salary"],
    it_employees,
    0)

print("Total IT Salary:", total_it_salary)


# 7. Print only employee names
names = list(
    map(lambda e: e["name"], employees))

print("Employee Names:", names)


# 8. Second highest salary
salary_sorted = sorted(
    employees,
    key=lambda e: e["salary"],
    reverse=True)

second_highest = salary_sorted[1]

print("Second Highest Salary Employee:", second_highest)
```

3. Product Store Analysis (★★★★★★)
products = [
{"name":"Laptop","price":65000,"rating":4.5},
{"name":"Mouse","price":800,"rating":4.1},
{"name":"Keyboard","price":1800,"rating":4.8},
{"name":"Monitor","price":14000,"rating":4.6},
{"name":"Tablet","price":28000,"rating":4.3},
{"name":"Phone","price":42000,"rating":4.9},
]
Using only lambda,
Find
1. Most expensive product
2. Cheapest product
3. Products above ₹20,000
4. Products with rating >4.5

Sumit Sir Notes
5. Sort by rating descending then price ascending
6. Average product price
7. Total inventory price
8. Print only product names sorted alphabetically


```python
from functools import reduce

products = [
    {"name": "Laptop", "price": 65000, "rating": 4.5},
    {"name": "Mouse", "price": 800, "rating": 4.1},
    {"name": "Keyboard", "price": 1800, "rating": 4.8},
    {"name": "Monitor", "price": 14000, "rating": 4.6},
    {"name": "Tablet", "price": 28000, "rating": 4.3},
    {"name": "Phone", "price": 42000, "rating": 4.9}
]


# 1. Most expensive product
most_expensive = max(
    products,
    key=lambda p: p["price"]
)

print("Most Expensive Product:", most_expensive)


# 2. Cheapest product
cheapest = min(
    products,
    key=lambda p: p["price"]
)

print("Cheapest Product:", cheapest)


# 3. Products above ₹20,000
above_20000 = list(
    filter(
        lambda p: p["price"] > 20000,
        products
    )
)

print(
    "Products Above 20000:",
    list(map(lambda p: p["name"], above_20000))
)


# 4. Products with rating > 4.5
high_rating = list(
    filter(
        lambda p: p["rating"] > 4.5,
        products
    )
)

print(
    "Products Rating > 4.5:",
    list(map(lambda p: p["name"], high_rating))
)


# 5. Sort by rating descending, then price ascending
sorted_products = sorted(
    products,
    key=lambda p: (-p["rating"], p["price"])
)

print(
    "Sorted Products:",
    list(
        map(
            lambda p: (p["name"], p["rating"], p["price"]),
            sorted_products
        )
    )
)


# 6. Average product price
total_price = reduce(
    lambda total, p: total + p["price"],
    products,
    0
)

average_price = total_price / len(products)

print("Average Product Price:", round(average_price, 2))


# 7. Total inventory price
print("Total Inventory Price:", total_price)


# 8. Product names sorted alphabetically
product_names = sorted(
    map(lambda p: p["name"], products)
)

print("Product Names:", product_names)
```

4. Customer Order Processing (★★★★★★)
orders = [
{"id":1,"customer":"Ram","amount":4500,"status":"Delivered"},
{"id":2,"customer":"Amit","amount":7000,"status":"Pending"},
{"id":3,"customer":"Raj","amount":2000,"status":"Delivered"},
{"id":4,"customer":"Neha","amount":9000,"status":"Delivered"},
{"id":5,"customer":"Anu","amount":3500,"status":"Cancelled"},
{"id":6,"customer":"Ravi","amount":10000,"status":"Delivered"},
]
Using only lambda,
1. Total delivered amount
2. Average delivered amount
3. Largest delivered order
4. Pending orders
5. Sort by amount descending
6. Print only delivered customers
7. Count delivered orders
8. Find customer spending maximum


```python
from functools import reduce

orders = [
    {"id": 1, "customer": "Ram", "amount": 4500, "status": "Delivered"},
    {"id": 2, "customer": "Amit", "amount": 7000, "status": "Pending"},
    {"id": 3, "customer": "Raj", "amount": 2000, "status": "Delivered"},
    {"id": 4, "customer": "Neha", "amount": 9000, "status": "Delivered"},
    {"id": 5, "customer": "Anu", "amount": 3500, "status": "Cancelled"},
    {"id": 6, "customer": "Ravi", "amount": 10000, "status": "Delivered"}]


# Get all delivered orders
delivered = list(
    filter(
        lambda o: o["status"] == "Delivered",
        orders))


# 1. Total delivered amount
total_delivered = reduce(
    lambda total, o: total + o["amount"],
    delivered,
    0)

print("Total Delivered Amount:", total_delivered)


# 2. Average delivered amount
average_delivered = total_delivered / len(delivered)

print("Average Delivered Amount:", average_delivered)


# 3. Largest delivered order
largest_delivered = max(
    delivered,
    key=lambda o: o["amount"])

print("Largest Delivered Order:", largest_delivered)


# 4. Pending orders
pending_orders = list(
    filter(
        lambda o: o["status"] == "Pending",
        orders))

print("Pending Orders:", pending_orders)


# 5. Sort all orders by amount descending
sorted_orders = sorted(
    orders,
    key=lambda o: o["amount"],
    reverse=True)

print(
    "Orders Sorted by Amount:",
    list(
        map(
            lambda o: (o["customer"], o["amount"]),
            sorted_orders)))


# 6. Print only delivered customers
delivered_customers = list(
    map(
        lambda o: o["customer"],
        delivered))

print("Delivered Customers:", delivered_customers)


# 7. Count delivered orders
delivered_count = len(delivered)

print("Delivered Order Count:", delivered_count)


# 8. Find customer spending maximum
maximum_order = max(
    orders,
    key=lambda o: o["amount"])

print("Maximum Spending Customer:", maximum_order["customer"])
print("Maximum Amount:", maximum_order["amount"])
```

```python
from functools import reduce

students = [{
        "name": "Ram",
        "age": 22,
        "marks": [80, 70, 90],
        "skills": ["Python", "SQL"]},{
        "name": "Amit",
        "age": 20,
        "marks": [95, 92, 88],
        "skills": ["Python", "Power BI", "SQL"]},{
        "name": "Neha",
        "age": 24,
        "marks": [85, 90, 80],
        "skills": ["Excel", "Python"]},{
        "name": "Raj",
        "age": 21,
        "marks": [65, 70, 72],
        "skills": ["SQL"]}]


# Function to calculate average of one student
average = lambda s: (
    reduce(lambda a, b: a + b, s["marks"], 0)
    /
    reduce(lambda count, x: count + 1, s["marks"], 0))


# 1. Calculate average marks
averages = reduce(
    lambda result, s:
        result + [(s["name"], average(s))],
    students,
    [])

print("1. Average Marks:")
print(averages)


# 2. Find topper
topper = max(
    students,
    key=lambda s: average(s))

print("\n2. Topper:")
print(topper["name"])


# 3. Find lowest scorer
lowest = min(
    students,
    key=lambda s: average(s))

print("\n3. Lowest Scorer:")
print(lowest["name"])


# 4. Sort by average descending
ranking = sorted(
    students,
    key=lambda s: average(s),
    reverse=True)

ranking_names = reduce(
    lambda result, s:
        result + [(s["name"], average(s))],
    ranking,
    [])

print("\n4. Ranking:")
print(ranking_names)


# 5. Names having Python skill
python_students = reduce(
    lambda result, s:
        result + [s["name"]]
        if any(map(lambda skill: skill == "Python", s["skills"]))
        else result,
    students,
    [])

print("\n5. Python Students:")
print(python_students)


# 6. Names having both Python and SQL
python_sql_students = reduce(
    lambda result, s:
        result + [s["name"]]
        if (
            any(map(lambda skill: skill == "Python", s["skills"]))
            and
            any(map(lambda skill: skill == "SQL", s["skills"]))
        )
        else result,
    students,
    [])

print("\n6. Python + SQL Students:")
print(python_sql_students)


# 7. Students older than 21
older_students = reduce(
    lambda result, s:
        result + [s["name"]]
        if s["age"] > 21
        else result,
    students,
    [])

print("\n7. Students Older Than 21:")
print(older_students)


# Count total students
total_students = reduce(
    lambda count, s: count + 1,
    students,
    0)


# 8. Overall average of all students
overall_average = (
    reduce(
        lambda total, s: total + average(s),
        students,
        0)/ total_students)

print("\n8. Overall Average:")
print(overall_average)


# Function to count skills
skill_count = lambda s: reduce(
    lambda count, skill: count + 1,
    s["skills"],
    0)


# 9. Student having maximum number of skills
most_skilled = max(
    students,
    key=lambda s: skill_count(s))

print("\n9. Maximum Skills:")
print(most_skilled["name"])


# 10. Names in alphabetical order
alphabetical_names = sorted(
    map(lambda s: s["name"], students))

print("\n10. Alphabetical Names:")
print(alphabetical_names)


# 11. Students above overall average
above_average = reduce(
    lambda result, s:
        result + [s["name"]]
        if average(s) > overall_average
        else result,
    students,
    [])

print("\n11. Above Overall Average:")
print(above_average)


# 12. Count students having Python
python_count = reduce(
    lambda count, s:
        count + 1
        if any(map(lambda skill: skill == "Python", s["skills"]))
        else count,
    students,
    0)

print("\n12. Python Student Count:")
print(python_count)


# 13. Count students having SQL
sql_count = reduce(
    lambda count, s:
        count + 1
        if any(map(lambda skill: skill == "SQL", s["skills"]))
        else count,
    students,
    0)

print("\n13. SQL Student Count:")
print(sql_count)


# 14. Print only names
all_names = reduce(
    lambda result, s:
        result + [s["name"]],
    students,
    [])

print("\n14. All Names:")
print(all_names)


# SQL student names for final dictionary
sql_students = reduce(
    lambda result, s:
        result + [s["name"]]
        if any(map(lambda skill: skill == "SQL", s["skills"]))
        else result,
    students,
    [])


# 15. Final dictionary
final_result = {
    "Topper": topper["name"],
    "Average": overall_average,
    "Python Students": python_students,
    "SQL Students": sql_students,
    "Above Average": above_average,
    "Total Students": total_students}

print("\n15. Final Dictionary:")
print(final_result)
```
