```python
#1. Create a list original = [1, 2, [3, 4], 5].
#Make a shallow copy named shallow and changeshallow[2][0] = 99.
#Print both lists and explain the result.
import copy

original = [1, 2, [3, 4], 5]

shallow = copy.copy(original)

shallow[2][0] = 99

print("Original:", original)
print("Shallow :", shallow)
```

```python
#2. Do the same as above but use deepcopy instead of shallow copy. Show the difference.
import copy

original = [1, 2, [3, 4], 5]

deep = copy.deepcopy(original)

deep[2][0] = 99

print("Original:", original)
print("Deep    :", deep)
```

```python
#3. Write a program to demonstrate shallow copy using copy.copy() on a nested list.
import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)

shallow[0][1] = 100

print("Original:", original)
print("Shallow :", shallow)
```

```python
#4. Write a program to demonstrate deep copy using copy.deepcopy() on the same nested list.
import copy

original = [[1, 2], [3, 4]]

deep = copy.deepcopy(original)

deep[0][1] = 100

print("Original:", original)
print("Deep    :", deep)
```

```python
#6. Create a class Person with a list attribute hobbies. 
#Create two objects using shallow copy and show how changing hobbies affects both.
import copy

class Person:

    def __init__(self, name, hobbies):
        self.name = name
        self.hobbies = hobbies


p1 = Person("Ram", ["Cricket", "Music"])

p2 = copy.copy(p1)

p2.hobbies.append("Reading")

print("Person 1:", p1.hobbies)
print("Person 2:", p2.hobbies)
```

```python
#7. Write a function manual_deep_copy(lst) that manually creates a deep copy of a nested list without using copy.deepcopy().
def manual_deep_copy(lst):

    result = []

    for item in lst:

        if isinstance(item, list):
            result.append(manual_deep_copy(item))
        else:
            result.append(item)

    return result


original = [1, [2, 3], [4, [5, 6]]]

copied = manual_deep_copy(original)

copied[1][0] = 99

print("Original:", original)
print("Copied  :", copied)
```

```python
#8. Given this data: data = [10, [20, 30], {"name": "Ram"}, 40] Make shallow and deep copies.
#Then change data[1][0] and data[2]["name"]. Print all three versions.
import copy

data = [
    10,
    [20, 30],
    {"name": "Ram"},
    40
]

shallow = copy.copy(data)
deep = copy.deepcopy(data)

data[1][0] = 99
data[2]["name"] = "Amit"

print("Original:", data)
print("Shallow :", shallow)
print("Deep    :", deep)
```

```python
#9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a nested element and checking the original.
import copy

def detect_copy(original, copied):

    old_value = original[0][0]

    copied[0][0] = 999

    if original[0][0] == 999:
        result = "Shallow Copy"
    else:
        result = "Deep Copy"

    copied[0][0] = old_value
    original[0][0] = old_value

    return result


original = [[1, 2], [3, 4]]

shallow = copy.copy(original)

print(detect_copy(original, shallow))


original = [[1, 2], [3, 4]]

deep = copy.deepcopy(original)

print(detect_copy(original, deep))
```

```python
#10. (Hard) Create a class Student with attributes name and marks (list).
#Demonstrate what happens when you do shallow copy vs deep copy of a Student object.
import copy

class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks


student1 = Student("Ram", [80, 90, 85])


# Shallow copy
student2 = copy.copy(student1)

student2.marks[0] = 100

print("After shallow copy:")
print(student1.marks)
print(student2.marks)


# Fresh object
student1 = Student("Ram", [80, 90, 85])


# Deep copy
student3 = copy.deepcopy(student1)

student3.marks[0] = 100

print("\nAfter deep copy:")
print(student1.marks)
print(student3.marks)
```

Withdraw should happen only if balance is sufficient.

```python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):

        if amount > 0:
            self.balance += amount
            print("Balance:", self.balance)
        else:
            print("Invalid deposit")

    def withdraw(self, amount):

        if amount <= 0:
            print("Invalid amount")

        elif amount <= self.balance:
            self.balance -= amount
            print("Remaining Balance:", self.balance)

        else:
            print("Insufficient balance")


account = BankAccount("Ram", 10000)

account.deposit(5000)
account.withdraw(3000)
```

```python
class CricketPlayer:

    def __init__(self, player_name, total_runs, matches):
        self.player_name = player_name
        self.total_runs = total_runs
        self.matches = matches

    def add_runs(self, run):

        if run < 0:
            print("Invalid runs")
            return

        self.total_runs += run

        if self.matches > 0:
            average = self.total_runs / self.matches
            print("Total Runs:", self.total_runs)
            print("Average:", average)
        else:
            print("Matches cannot be zero")


player = CricketPlayer("Virat", 5000, 100)

player.add_runs(100)
```

23. Student Attendance

mark_attendance(number)

Print attendance percentage after every update.

```python
class StudentAttendance:

    def __init__(self, total_classes, attended_classes):
        self.total_classes = total_classes
        self.attended_classes = attended_classes

    def mark_attendance(self, number):

        if number <= 0:
            print("Invalid number")

        elif self.attended_classes + number <= self.total_classes:

            self.attended_classes += number

            percentage = (
                self.attended_classes /
                self.total_classes
            ) * 100

            print("Attended:", self.attended_classes)
            print("Attendance:", round(percentage, 2), "%")

        else:
            print("Attendance cannot exceed total classes")


student = StudentAttendance(100, 70)

student.mark_attendance(10)
```

24. Bus Reservation (Hard)

• booked_seats
Method
reserve(number)
Reservation allowed only if
booked_seats + number
does not exceed total seats.
After booking print
Remaining seats.

```python
class BusReservation:

    def __init__(self, bus_name, total_seats, booked_seats):
        self.bus_name = bus_name
        self.total_seats = total_seats
        self.booked_seats = booked_seats

    def reserve(self, number):

        if number <= 0:
            print("Invalid number of seats")

        elif self.booked_seats + number <= self.total_seats:

            self.booked_seats += number

            remaining = (
                self.total_seats -
                self.booked_seats
            )

            print("Reservation Successful")
            print("Remaining Seats:", remaining)

        else:
            print("Not enough seats")


bus = BusReservation("Express", 50, 30)

bus.reserve(10)
```

25. Online Shopping Cart (Interview Level)
Attributes
• customer_name
• cart_total
• wallet_balance
Methods
1. add_item(price)
2. remove_item(price)
3. checkout()
Rules
• Cart total cannot become negative.
• Checkout only if wallet balance is enough.
• Deduct wallet balance after successful checkout.
• Cart becomes zero after payment.
• Print suitable messages for every case.

```python
class ShoppingCart:

    def __init__(self, customer_name, cart_total, wallet_balance):
        self.customer_name = customer_name
        self.cart_total = cart_total
        self.wallet_balance = wallet_balance

    def add_item(self, price):

        if price <= 0:
            print("Invalid price")

        else:
            self.cart_total += price
            print("Item added")
            print("Cart Total:", self.cart_total)

    def remove_item(self, price):

        if price <= 0:
            print("Invalid price")

        elif price <= self.cart_total:
            self.cart_total -= price
            print("Item removed")
            print("Cart Total:", self.cart_total)

        else:
            print("Cart total cannot become negative")

    def checkout(self):

        if self.cart_total == 0:
            print("Cart is empty")

        elif self.wallet_balance >= self.cart_total:

            self.wallet_balance -= self.cart_total

            print("Payment successful")

            self.cart_total = 0

            print("Wallet Balance:", self.wallet_balance)
            print("Cart Total:", self.cart_total)

        else:
            print("Insufficient wallet balance")


customer = ShoppingCart("Ram", 0, 10000)

customer.add_item(3000)
customer.add_item(2000)
customer.remove_item(1000)
customer.checkout()
```

Bonus Challenge (Very Hard)
Create a University Student Management System.
Each student has
• Student ID
• Name
• Age
• Course
• Fees
• Marks
• Attendance
Create methods to
• Update marks (0–100 only)
• Pay fees (cannot exceed remaining fee)
• Update attendance (cannot exceed total classes)
• Change course
• Display student report
• Calculate attendance percentage
• Check Pass/Fail (Marks ≥ 35)
• Give Scholarship
o Marks > 90
o Attendance > 95%
o Fee completely paid

Print
• Student Report Card
• Fee Status
• Attendance %
• Scholarship Eligible (Yes/No)

```python
class UniversityStudent:

    def __init__(
        self,
        student_id,
        name,
        age,
        course,
        total_fee,
        paid_fee,
        marks,
        total_classes,
        attended_classes
    ):

        self.student_id = student_id
        self.name = name
        self.age = age
        self.course = course
        self.total_fee = total_fee
        self.paid_fee = paid_fee
        self.marks = marks
        self.total_classes = total_classes
        self.attended_classes = attended_classes


    # Update marks
    def update_marks(self, new_marks):

        if 0 <= new_marks <= 100:
            self.marks = new_marks
            print("Marks updated:", self.marks)

        else:
            print("Invalid marks")


    # Pay fees
    def pay_fees(self, amount):

        remaining = self.total_fee - self.paid_fee

        if amount <= 0:
            print("Invalid amount")

        elif amount <= remaining:

            self.paid_fee += amount

            print("Payment successful")
            print(
                "Remaining Fee:",
                self.total_fee - self.paid_fee
            )

        else:
            print("Amount exceeds remaining fee")


    # Update attendance
    def update_attendance(self, number):

        if number < 0:
            print("Invalid attendance")

        elif self.attended_classes + number <= self.total_classes:

            self.attended_classes += number

            print(
                "Attended Classes:",
                self.attended_classes
            )

        else:
            print("Attendance cannot exceed total classes")


    # Change course
    def change_course(self, new_course):

        self.course = new_course

        print(
            "Course changed to:",
            self.course
        )


    # Attendance percentage
    def attendance_percentage(self):

        if self.total_classes == 0:
            return 0

        return (
            self.attended_classes /
            self.total_classes
        ) * 100


    # Pass or fail
    def check_result(self):

        if self.marks >= 35:
            return "Pass"

        return "Fail"


    # Scholarship
    def scholarship(self):

        attendance = self.attendance_percentage()

        fee_paid = (
            self.paid_fee == self.total_fee
        )

        if (
            self.marks > 90
            and attendance > 95
            and fee_paid
        ):
            return "Yes"

        return "No"


    # Fee status
    def fee_status(self):

        remaining = (
            self.total_fee -
            self.paid_fee
        )

        if remaining == 0:
            return "Fully Paid"

        return "Remaining Fee: " + str(remaining)


    # Display complete report
    def display_report(self):

        print("=" * 40)
        print("STUDENT REPORT CARD")
        print("=" * 40)

        print("Student ID:", self.student_id)
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)

        print("Marks:", self.marks)
        print("Result:", self.check_result())

        print(
            "Attendance:",
            round(self.attendance_percentage(), 2),
            "%"
        )

        print(
            "Fee Status:",
            self.fee_status()
        )

        print(
            "Scholarship Eligible:",
            self.scholarship()
        )

        print("=" * 40)


student = UniversityStudent(
    student_id=101,
    name="Amit",
    age=20,
    course="Python Data Science",
    total_fee=100000,
    paid_fee=100000,
    marks=95,
    total_classes=100,
    attended_classes=97
)


student.display_report()
```
