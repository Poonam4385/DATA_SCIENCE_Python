String Slicing Questions
1. Take a string from user and print the first 5 characters.
2. Take a string and print the last 4 characters using negative slicing.
3. Take a string and print characters from index 2 to 8 (excluding 8).
4. Take a string and print every 2nd character of the string.
5. Take a full name (e.g. "Sumit Kumar Sharma") and print only the first
name and last name separately using slicing.

```python
string = input("string:")
print("frist 5 charatore:",string[:5])
print("last 4 charatore:",string[-4:])
print("index 2 to 8:",string[2:8])
print("every 2nd charactore:",string[::2])
print("frist name:",string[:string.index(" ")])
print("last name:",string.split()[-1])
```

### String Check Methods Questions
9. Take a string from user and check whether it contains only alphabets
(isalpha()).
10. Take input from user and check if it is a digit only (isdigit()).
11. Take a string and check if it is alphanumeric (isalnum()).
12. Take a string and check if it is completely in lowercase (islower()).
13. Take a password as input and check:
• It should be alphanumeric
• It should be minimum 8 characters long Print "Valid Password" or
"Invalid Password".

```python
string =input("string:")

if string.isalpha():
    print("contains only alphabets")
else:
    print("not contains only alphabets")

if string.isdigit():
    print("digit only")
else:
     print("no digit only")


if string.isalnum():
    print("alphanumeric")
else:
    print("no alphanumeric")

if string.islower():
    print("completely in lowercase")
else:
     print("no lowercase")
```

```python
password = input("Pssword:")

if password.isalnum() and len(password)>= 8:
    print("Valid Password")
else:
    print("Invalid Password")
    

```
