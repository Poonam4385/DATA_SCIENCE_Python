### Method Overloading Questions (1-7)

1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method
name.
Concept
Method Overloading (Python uses default arguments or *args to achieve overloading.)
What to do
• Create a class Calculator.
• Create one method add().
• The method should work for either 2 numbers or 3 numbers.
How to solve
1. Create class Calculator.
2. Define method add().
3. Use default arguments or *args.
4. Check how many numbers are provided.
5. Return the sum.

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

obj = Calculator()

print(obj.add(10, 20))
print(obj.add(10, 20, 30))
```

2. Create an Area class that calculates the area of a square and rectangle using the same
method.
Concept
Method Overloading using default arguments.
What to do
• One parameter → Square.
• Two parameters → Rectangle.
How to solve
1. Create class Area.
2. Define method calculate().
3. If only one value is provided, calculate the square's area.
4. If two values are provided, calculate the rectangle's area.
5. Return the result.

```python
class Area:
    def calculate(self, length, width=None):
        if width is None:
            return length * length
        else:
            return length * width

obj = Area()

print("Square Area:", obj.calculate(5))
print("Rectangle Area:", obj.calculate(5, 4))
```

3. Create a Student class with a details() method that accepts different numbers of
arguments.
Concept
Method Overloading
What to do
The method should work with:
• Name
• Name + Age
• Name + Age + City
How to solve
1. Create class Student.
2. Create method details().
3. Use default arguments.
4. Print the information based on the provided values.

```python
class Student:
    def details(self, name, age=None, city=None):
        print("Name:", name)

        if age is not None:
            print("Age:", age)

        if city is not None:
            print("City:", city)

obj = Student()

obj.details("Rahul")
obj.details("Rahul", 20)
obj.details("Rahul", 20, "Pune")
```

4. Create a MathOperations class that multiplies any number of values.
Concept
Method Overloading using *args
What to do
Multiply:
• 2 numbers
• 3 numbers
• 4 numbers
• More numbers
How to solve
1. Create class MathOperations.
2. Define method multiply(*args).
3. Initialize the answer as 1.
4. Loop through all numbers.
5. Multiply each number.
6. Return the final product.

```python
class MathOperations:
    def multiply(self, *args):
        result = 1

        for num in args:
            result *= num

        return result

obj = MathOperations()

print(obj.multiply(2, 3))
print(obj.multiply(2, 3, 4))
print(obj.multiply(2, 3, 4, 5))
```

5. Create a Display class that prints different messages based on the argument type.
Concept
Method Overloading + Type Checking
What to do
Accept:
• Integer
• String
• List
How to solve
1. Create class Display.
2. Create method show().
3. Use isinstance().
4. Print different messages depending on the object's type.

```python
class Display:
    def show(self, value):
        if isinstance(value, int):
            print("Integer:", value)

        elif isinstance(value, str):
            print("String:", value)

        elif isinstance(value, list):
            print("List:", value)

        else:
            print("Unknown type")

obj = Display()

obj.show(100)
obj.show("Python")
obj.show([10, 20, 30])
```

6. Create a Volume class that calculates the volume of a cube and cuboid.
Concept
Method Overloading
What to do
• One parameter → Cube
• Three parameters → Cuboid
How to solve
1. Create class Volume.
2. Create method calculate().
3. If one value is given, calculate the cube's volume.
4. Otherwise, calculate the cuboid's volume.

```python
class Volume:
    def calculate(self, length, width=None, height=None):
        if width is None and height is None:
            return length ** 3
        else:
            return length * width * height

obj = Volume()

print("Cube Volume:", obj.calculate(5))
print("Cuboid Volume:", obj.calculate(5, 4, 3))
```

7. Create a PrintData class that prints one or many values.
Concept
Method Overloading using *args
What to do
Print:
• One value
• Multiple values
How to solve
1. Create class PrintData.
2. Create method print_data(*args).
3. Loop through the arguments.
4. Print each value.

8. Create an Animal class with sound(). Override it in Dog and Cat.
Concept
Method Overriding
What to do
• Animal → Generic sound
• Dog → Bark
• Cat → Meow
How to solve
1. Create parent class Animal.
2. Define method sound().
3. Inherit Dog and Cat.
4. Override sound().
5. Create objects and call the method.

```python
class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def sound(self):
        print("Dog says Bark")

class Cat(Animal):
    def sound(self):
        print("Cat says Meow")

d = Dog()
c = Cat()

d.sound()
c.sound()
```

10. Create an Employee class and override calculate_salary() in Manager.
Concept
Method Overriding
What to do
• Employee → Normal salary
• Manager → Salary + Bonus
How to solve
1. Create parent class.
2. Define calculate_salary().
3. Inherit Manager.
4. Override the method.
5. Print the manager's salary.

```python
class Employee:
    def __init__(self, salary):
        self.salary = salary

    def calculate_salary(self):
        return self.salary

class Manager(Employee):
    def __init__(self, salary, bonus):
        super().__init__(salary)
        self.bonus = bonus

    def calculate_salary(self):
        return self.salary + self.bonus

e = Employee(50000)
m = Manager(50000, 10000)

print("Employee Salary:", e.calculate_salary())
print("Manager Salary:", m.calculate_salary())
```

11. Create Shape, Circle, and Rectangle classes.
Concept
Method Overriding
What to do
Override the area() method.
How to solve
1. Create Shape.
2. Add area().
3. Inherit Circle.
4. Inherit Rectangle.
5. Override area() in each child class.

```python
import math

class Shape:
    def area(self):
        print("Area not defined")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(10, 4)

print("Circle Area:", c.area())
print("Rectangle Area:", r.area())
```

##### Operator Overloading Questions (15-20)

15. Create a Point class and overload the + operator.
Concept
Operator Overloading (__add__)
What to do
Add two points together.
How to solve
1. Create class Point.
2. Store x and y coordinates.
3. Overload __add__().
4. Return a new point with added coordinates.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print(self.x, self.y)

p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

p3.display()
```

18. Create a Rectangle class and overload the == operator.
Concept
Equality Operator Overloading (__eq__)
What to do
Compare rectangles using their area.
How to solve
1. Create class Rectangle.
2. Store length and width.
3. Calculate area.
4. Overload __eq__().
5. Compare the areas.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def __eq__(self, other):
        return self.area() == other.area()

r1 = Rectangle(5, 4)
r2 = Rectangle(10, 2)

print(r1 == r2)
```

19. Create a Money class and overload the + and - operators.
Concept
Multiple Operator Overloading
What to do
Perform addition and subtraction of money amounts.
How to solve
1. Create class Money.
2. Store the amount.
3. Overload __add__().
4. Overload __sub__().
5. Return new Money objects after each operation.

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __sub__(self, other):
        return Money(self.amount - other.amount)

    def display(self):
        print("Amount:", self.amount)

m1 = Money(5000)
m2 = Money(2000)

m3 = m1 + m2
m4 = m1 - m2

m3.display()
m4.display()
```

20. Create a Distance class and overload <, >, and ==.
Concept
Comparison Operator Overloading
What to do
Compare two distance objects.
How to solve
1. Create class Distance.
2. Store the distance value.
3. Overload:
o __lt__() for <
o __gt__() for >
o __eq__() for ==
4. Compare the stored distance values.
5. Return True or False.

```python
class Distance:
    def __init__(self, value):
        self.value = value

    def __lt__(self, other):
        return self.value < other.value

    def __gt__(self, other):
        return self.value > other.value

    def __eq__(self, other):
        return self.value == other.value

d1 = Distance(100)
d2 = Distance(200)

print("d1 < d2 :", d1 < d2)
print("d1 > d2 :", d1 > d2)
print("d1 == d2:", d1 == d2)
```
