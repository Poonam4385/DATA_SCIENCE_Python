1. Class Name: Shape
Problem
Create an abstract class named Shape.
Explanation
A shape can have different formulas for area and perimeter. Therefore, create two abstract
methods:
• area()
• perimeter()
Create subclasses like Circle and Rectangle and implement both methods.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


c = Circle(5)
r = Rectangle(10, 4)

print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())

print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())
```

15. Class Name: Shape3D
Problem
Create an abstract class named Shape3D.
Explanation
Three-dimensional shapes have different formulas.
Create abstract methods:
• volume()
• surface_area()
Create subclasses:
• Cube
• Sphere

```python
from abc import ABC, abstractmethod
import math

class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass


class Cube(Shape3D):

    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * self.side ** 2


class Sphere(Shape3D):

    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4 / 3) * math.pi * self.radius ** 3

    def surface_area(self):
        return 4 * math.pi * self.radius ** 2


cube = Cube(5)
sphere = Sphere(5)

print("Cube Volume:", cube.volume())
print("Cube Surface Area:", cube.surface_area())

print("Sphere Volume:", sphere.volume())
print("Sphere Surface Area:", sphere.surface_area())
```

Class Name: BankAccount
Problem
Create an abstract class named BankAccount.
Explanation
A bank account stores the account balance, but different account types calculate interest
differently.
Create a constructor:
__init__(self, balance)
Store the balance as an instance variable.
Create an abstract method:
calculate_interest()
Create the following subclasses
SavingsAccount
• Use super() to initialize the balance.
• Implement calculate_interest().
• The account should maintain a minimum balance.
• If the balance falls below the minimum balance, display an appropriate message or
prevent the operation.
CurrentAccount
• Use super() to initialize the balance.
• Implement calculate_interest().
• Allow an overdraft, meaning the balance is allowed to become negative up to a
specified limit.
Additional Requirements
• Create a class variable to count the total number of bank accounts created.
• Create a class method named show_total_accounts() to display the total number of
accounts.
• Demonstrate creating multiple account objects and calling the class method.
Concepts Covered
• Abstract Class
• Constructor
• super()
• Method Overriding
• Class Variable
• Class Method
• Validation

```python
from abc import ABC, abstractmethod

class BankAccount(ABC):

    total_accounts = 0

    def __init__(self, balance):
        self.balance = balance
        BankAccount.total_accounts += 1

    @abstractmethod
    def calculate_interest(self):
        pass

    @classmethod
    def show_total_accounts(cls):
        print("Total Accounts:", cls.total_accounts)


class SavingsAccount(BankAccount):

    def __init__(self, balance, minimum_balance=1000):
        super().__init__(balance)
        self.minimum_balance = minimum_balance

    def calculate_interest(self):
        return self.balance * 0.04

    def withdraw(self, amount):

        if amount <= 0:
            print("Invalid amount")

        elif self.balance - amount < self.minimum_balance:
            print("Minimum balance must be maintained")

        else:
            self.balance -= amount
            print("Balance:", self.balance)


class CurrentAccount(BankAccount):

    def __init__(self, balance, overdraft_limit=5000):
        super().__init__(balance)
        self.overdraft_limit = overdraft_limit

    def calculate_interest(self):
        return self.balance * 0.02 if self.balance > 0 else 0

    def withdraw(self, amount):

        if amount <= 0:
            print("Invalid amount")

        elif self.balance - amount < -self.overdraft_limit:
            print("Overdraft limit exceeded")

        else:
            self.balance -= amount
            print("Balance:", self.balance)


s1 = SavingsAccount(10000)
s2 = SavingsAccount(20000)
c1 = CurrentAccount(5000)

print("Savings Interest:", s1.calculate_interest())

s1.withdraw(5000)

c1.withdraw(8000)

BankAccount.show_total_accounts()
```

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    def __init__(self, color):
        self.color = color

    def display_color(self):
        print("Color:", self.color)

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Rectangle(Shape):

    def __init__(self, color, length, width):
        super().__init__(color)
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def display_details(self):
        super().display_color()
        print("Length:", self.length)
        print("Width:", self.width)
        print("Area:", self.area())
        print("Perimeter:", self.perimeter())


class Circle(Shape):

    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

    def display_details(self):
        super().display_color()
        print("Radius:", self.radius)
        print("Area:", self.area())
        print("Perimeter:", self.perimeter())


r = Rectangle("Blue", 10, 5)
c = Circle("Red", 5)

r.display_details()
c.display_details()
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    def __init__(self, brand, fuel_type):
        self.brand = brand
        self.fuel_type_name = fuel_type

    def display_vehicle(self):
        print("Brand:", self.brand)
        print("Fuel Type:", self.fuel_type_name)

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def fuel_consumption(self, distance):
        pass


class ElectricCar(Vehicle):

    def __init__(self, brand, fuel_type, km_per_kwh):
        super().__init__(brand, fuel_type)

        if fuel_type != "Electric":
            print("Invalid fuel type for Electric Car")

        self.km_per_kwh = km_per_kwh

    def start(self):
        print("Electric car starts silently")

    def fuel_consumption(self, distance):
        return distance / self.km_per_kwh

    def display_vehicle(self):
        super().display_vehicle()
        print("Efficiency:", self.km_per_kwh, "km/kWh")


class PetrolCar(Vehicle):

    def __init__(self, brand, fuel_type, km_per_litre):
        super().__init__(brand, fuel_type)

        if fuel_type != "Petrol":
            print("Invalid fuel type for Petrol Car")

        self.km_per_litre = km_per_litre

    def start(self):
        print("Petrol car starts with engine")

    def fuel_consumption(self, distance):
        return distance / self.km_per_litre

    def display_vehicle(self):
        super().display_vehicle()
        print("Mileage:", self.km_per_litre, "km/L")


electric = ElectricCar(
    "Tesla",
    "Electric",
    6
)

petrol = PetrolCar(
    "Toyota",
    "Petrol",
    15
)

electric.display_vehicle()
electric.start()
print(
    "Battery Used:",
    electric.fuel_consumption(120),
    "kWh"
)

petrol.display_vehicle()
petrol.start()
print(
    "Petrol Used:",
    petrol.fuel_consumption(120),
    "litres"
)
```

Class Name: Report
Problem
Create an abstract class named Report.
Explanation
Every report has common information like title and data, but different report formats
generate output differently.
Create a constructor:
__init__(self, title, data)
Store:
• Report Title
• Report Data
Create two normal methods:
header()
Displays the report title.
footer()
Displays a closing message indicating the report has ended.
Create one abstract method:
generate()


Create the following subclasses
PDFReport
• Use super() in the constructor.
• Inside generate():
o Call super().header()
o Generate the PDF report
o Call super().footer()

HTMLReport
• Use super() in the constructor.
• Inside generate():
o Call super().header()
o Generate the HTML report
o Call super().footer()

Concepts Covered
• Abstract Class
• Constructor
• super()
• Parent Methods
• Method Overriding
• Code Reusability
• Real-world Report Generation

```python
from abc import ABC, abstractmethod

class Report(ABC):

    def __init__(self, title, data):
        self.title = title
        self.data = data

    def header(self):
        print("=" * 40)
        print(self.title)
        print("=" * 40)

    def footer(self):
        print("=" * 40)
        print("Report Ended")
        print("=" * 40)

    @abstractmethod
    def generate(self):
        pass


class PDFReport(Report):

    def __init__(self, title, data):
        super().__init__(title, data)

    def generate(self):

        super().header()

        print("Generating PDF Report")
        print("Data:", self.data)

        super().footer()


class HTMLReport(Report):

    def __init__(self, title, data):
        super().__init__(title, data)

    def generate(self):

        super().header()

        print("Generating HTML Report")
        print("Data:", self.data)

        super().footer()


pdf = PDFReport(
    "Sales Report",
    "Total Sales = 500000"
)

html = HTMLReport(
    "Employee Report",
    "Total Employees = 100"
)

pdf.generate()
html.generate()
```
