### Hybrid Inheritance Questions (1-10)

1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
from both B and C).
Concept
Hybrid Inheritance (Combination of Single + Multiple Inheritance)
What to do
• Create parent class A.
• Create class B inheriting from A.
• Create another class C.
• Create class D inheriting from both B and C.
How to solve
1. Create class A.
2. Add one method.
3. Create class B(A).
4. Create class C.
5. Create class D(B, C).
6. Create an object of D.
7. Call methods from all parent classes.

```python
class A:
    def show_a(self):
        print("Class A")

class B(A):
    def show_b(self):
        print("Class B")

class C:
    def show_c(self):
        print("Class C")

class D(B, C):
    def show_d(self):
        print("Class D")

obj = D()

obj.show_a()
obj.show_b()
obj.show_c()
obj.show_d()
```

2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
Concept
Real-world Hybrid Inheritance
What to do
• Person is the base class.
• Employee inherits Person.
• Student inherits Person.
• TeachingAssistant inherits Employee and Student.
How to solve

Sumit Sir Notes
Hierarchy
Person
/ \
Employee Student
\ /
TeachingAssistant
Steps
1. Store name in Person.
2. Employee stores salary.
3. Student stores marks.
4. TeachingAssistant stores department.
5. Display all information.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        super().__init__(name)
        self.salary = salary

class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        Person.__init__(self, name)
        self.salary = salary
        self.marks = marks
        self.department = department

    def display(self):
        print(self.name)
        print(self.salary)
        print(self.marks)
        print(self.department)

ta = TeachingAssistant("Rahul", 50000, 90, "CS")
ta.display()
```

Concept

Sumit Sir Notes
Diamond Problem
Hierarchy
A

B C

D
What to do
Use super() everywhere.
How to solve
1. Create constructor in A.
2. Call super() in B.
3. Call super() in C.

6. Observe A constructor runs only once.

```python
class A:
    def __init__(self):
        print("A Constructor")
        super().__init__()

class B(A):
    def __init__(self):
        print("B Constructor")
        super().__init__()

class C(A):
    def __init__(self):
        print("C Constructor")
        super().__init__()

class D(B, C):
    def __init__(self):
        print("D Constructor")
        super().__init__()

obj = D()
```

6. Create a banking system using hybrid inheritance.
Concept
Hybrid + Real-world Example
What to do
Account
|
Savings

Loan

SavingsLoanAccount
How to solve
1. Account stores account number.
2. Savings stores balance.
3. Loan stores loan amount.
4. SavingsLoanAccount inherits both.
5. Print all details.

```python
class Account:
    def __init__(self, account_number):
        self.account_number = account_number

class Savings(Account):
    def __init__(self, account_number, balance):
        super().__init__(account_number)
        self.balance = balance

class Loan:
    def __init__(self, loan_amount):
        self.loan_amount = loan_amount

class SavingsLoanAccount(Savings, Loan):
    def __init__(self, account_number, balance, loan_amount):
        Savings.__init__(self, account_number, balance)
        Loan.__init__(self, loan_amount)

    def display(self):
        print(self.account_number)
        print(self.balance)
        print(self.loan_amount)

obj = SavingsLoanAccount(101, 50000, 200000)
obj.display()
```

7. Create Shape, Color, Circle, and ColoredCircle.
Concept

What to do

Color

How to solve
1. Shape has area().
2. Color has color().
3. Circle inherits Shape.
4. ColoredCircle inherits Circle and Color.
5. Display area and color.

```python
import math

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Color:
    def color(self):
        return "Red"

class ColoredCircle(Circle, Color):
    pass

c = ColoredCircle(5)

print(c.area())
print(c.color())
```

15. Create Circle and Rectangle both having area().
Concept
Duck Typing
What to do
Function
calculate(obj)
How to solve
1. Circle implements area().
2. Rectangle implements area().
3. Function calls area().
4. Print returned value.

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

def calculate(obj):
    print(obj.area())

calculate(Circle(5))
calculate(Rectangle(10, 4))
```

20. Create Employee hierarchy using Hybrid Inheritance and use Duck Typing.
Concept
Hybrid Inheritance + Duck Typing
What to do
Hierarchy
Person
|
Employee

Manager

HR

TeamLead
Another class
Freelancer
Both TeamLead and Freelancer have
work()
How to solve
1. Build hybrid inheritance.
2. Final child defines work().
3. Freelancer also defines work().
4. Write a function:
def do_work(obj):
obj.work()
5. Pass both objects to the function.
6. Observe that the same function works for unrelated classes because of Duck Typing.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def employee_info(self):
        print(self.name, "is an employee")

class Manager(Employee):
    def manage(self):
        print(self.name, "manages team")

class HR:
    def recruit(self):
        print("Recruiting employees")

class TeamLead(Manager, HR):
    def work(self):
        print(self.name, "works as Team Lead")

class Freelancer:
    def __init__(self, name):
        self.name = name

    def work(self):
        print(self.name, "works as Freelancer")

def do_work(obj):
    obj.work()

t = TeamLead("Rahul")
f = Freelancer("Amit")

do_work(t)
do_work(f)
```
