10. Keep taking numbers from user until they enter 0, then print sum.

11. Reverse a number using while loop.
Explanation: Use modulo and division to reverse digits.
Input: 1234
Output: 4321

```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse =", reverse)
```

#12. Check if a number is prime using while loop.
#Explanation: Check divisibility from 2 to sqrt(n).Input: 17Output: Prime Number
num=1

```python
n = int(input("Enter a number: "))

if n < 2:
    print("Not Prime Number")
else:
    i = 2
    is_prime = True

    while i * i <= n:
        if n % i == 0:
            is_prime = False
            break

        i = i + 1

    if is_prime:
        print("Prime Number")
    else:
        print("Not Prime Number")
```

13. Print Fibonacci series up to n terms using while loop.
Explanation: Use two variables a=0, b=1 and generate next terms.
Input: 8
Output: 0 1 1 2 3 5 8 13

```python
n = int(input("Enter number of terms: "))

a = 0
b = 1
count = 0

while count < n:
    print(a, end=" ")

    c = a + b
    a = b
    b = c

    count = count + 1
```

22. Print a hollow square pattern of size n using nested loop.
Input: 5
Output: Square with only border *

```python
n = int(input("Enter size: "))

for i in range(n):

    for j in range(n):

        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

33. Print prime numbers between 10 to 50, use break and continue.
Input: None Output: Primes in range

```python
for num in range(10, 51):

    if num < 2:
        continue

    is_prime = True

    for i in range(2, num):

        if num % i == 0:
            is_prime = False
            break

    if not is_prime:
        continue

    print(num, end=" ")
```
