#### Basic Multiple Inheritance (1-10)

```python
#1. Create two parent classes Father and Mother. Create a child class Child that inherits from both.
class Father:
    def father(self):
        print("Father class")

class Mother:
    def mother(self):
        print("Mother class")

class Child(Father, Mother):
    pass

obj = Child()
obj.father()
obj.mother()
```

### Multiple Inheritance with Static Methods (11-18)

```python
#11. Create two parent classes with static methods. Create a child class that inherits from both and calls the static methods.
class A:
    @staticmethod
    def method1():
        print("Static method A")

class B:
    @staticmethod
    def method2():
        print("Static method B")

class C(A, B):
    pass

C.method1()
C.method2()
```

```python
#12. Define class MathUtils with static method add(). Create class AdvancedMath thatinherits and uses the static method.
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

class AdvancedMath(MathUtils):
    pass

print(AdvancedMath.add(10, 20))
```

```python
#13. Create class Validator with static method is_valid(). Create class Form that inherits and uses it.
class Validator:
    @staticmethod
    def is_valid(value):
        return value != ""

class Form(Validator):
    pass

print(Form.is_valid("Python"))
print(Form.is_valid(""))
```

```python
#14. Define class Converter with static method to_celsius(). Create class WeatherApp that inherits from multiple classes including Converter.
class Converter:
    @staticmethod
    def to_celsius(f):
        return (f - 32) * 5 / 9

class Weather:
    def information(self):
        print("Weather information")

class WeatherApp(Converter, Weather):
    pass

print(WeatherApp.to_celsius(98.6))
```

```python
#15. Create class Logger with static method log(). Create class Application that inherits from Logger and another class.
class Logger:
    @staticmethod
    def log(message):
        print("LOG:", message)

class Database:
    def connect(self):
        print("Database connected")

class Application(Logger, Database):
    pass

app = Application()
app.log("Application started")
app.connect()
```

#### Multiple Inheritance with Abstract Methods (19-25)

```python
#19. Create two abstract classes Shape and Color. Create a concrete class that inherits fromboth and implements abstract methods.
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Color(ABC):
    @abstractmethod
    def color(self):
        pass

class Circle(Shape, Color):
    def area(self):
        print("Area = 78.5")

    def color(self):
        print("Color = Red")

c = Circle()
c.area()
c.color()
```

```python
#20. Define abstract class Payment and abstract class Notification. Create class OnlinePayment that inherits from both.
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass

class Notification(ABC):
    @abstractmethod
    def notify(self):
        pass

class OnlinePayment(Payment, Notification):
    def pay(self):
        print("Payment completed")

    def notify(self):
        print("Notification sent")

obj = OnlinePayment()
obj.pay()
obj.notify()
```

```python
#21. Create abstract class Vehicle and abstract class Electric. Create class ElectricCar that inherits from both.
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Electric(ABC):
    @abstractmethod
    def charge(self):
        pass

class ElectricCar(Vehicle, Electric):
    def start(self):
        print("Car started")

    def charge(self):
        print("Battery charging")

car = ElectricCar()
car.start()
car.charge()
```

```python
#22. Define abstract class Readable and abstract class Writable. Create class File that implements both.
from abc import ABC, abstractmethod

class Readable(ABC):
    @abstractmethod
    def read(self):
        pass

class Writable(ABC):
    @abstractmethod
    def write(self):
        pass

class File(Readable, Writable):
    def read(self):
        print("Reading file")

    def write(self):
        print("Writing file")

f = File()
f.read()
f.write()
```

```python
#23. Create abstract class Flyable and abstract class Swimmable. Create class Duck thatinherits from both
from abc import ABC, abstractmethod

class Flyable(ABC):
    @abstractmethod
    def fly(self):
        pass

class Swimmable(ABC):
    @abstractmethod
    def swim(self):
        pass

class Duck(Flyable, Swimmable):
    def fly(self):
        print("Duck is flying")

    def swim(self):
        print("Duck is swimming")

d = Duck()
d.fly()
d.swim()
```

```python
#24. Demonstrate multiple inheritance where one parent is abstract and other is normalclass.
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Pet:
    def play(self):
        print("Pet playing")

class Dog(Animal, Pet):
    def sound(self):
        print("Woof")

d = Dog()
d.sound()
d.play()
```

```python
#25. Create two abstract classes with abstract methods of same name. Resolve in the childclass.
from abc import ABC, abstractmethod

class A(ABC):
    @abstractmethod
    def show(self):
        pass

class B(ABC):
    @abstractmethod
    def show(self):
        pass

class C(A, B):
    def show(self):
        print("Show implemented in C")

obj = C()
obj.show()
```

26. (Problem) Create two parent classes:
• Employee with method get_salary()
• Person with method get_details() Create class Manager that inherits from both. Add
a method that combines both parent methods.

```python
class Employee:
    def get_salary(self):
        return 50000

class Person:
    def get_details(self):
        return "Name: Rahul"

class Manager(Employee, Person):
    def details(self):
        print(self.get_details())
        print("Salary:", self.get_salary())

m = Manager()
m.details()
```

27. (Problem) Create abstract class Device with abstract method power_on() and class
Network with method connect(). Create class SmartPhone that inherits from both and
implements the abstract method.

```python
from abc import ABC, abstractmethod

class Device(ABC):
    @abstractmethod
    def power_on(self):
        pass

class Network:
    def connect(self):
        print("Network connected")

class SmartPhone(Device, Network):
    def power_on(self):
        print("SmartPhone ON")

phone = SmartPhone()
phone.power_on()
phone.connect()
```

28. (Problem) Design a system using multiple inheritance:
• Class Writer
• Class Editor Create class Author that inherits from both. Add static method to count
total authors.

```python
class Writer:
    def write(self):
        print("Writing article")

class Editor:
    def edit(self):
        print("Editing article")

class Author(Writer, Editor):

    count = 0

    def __init__(self, name):
        self.name = name
        Author.count += 1

    @staticmethod
    def total_authors():
        return Author.count

a1 = Author("Rahul")
a2 = Author("Amit")
a3 = Author("Priya")

a1.write()
a1.edit()

print("Total Authors:", Author.total_authors())
```

29. (Problem) Create two abstract classes Printable and Scannable. Create class
PrinterScanner that inherits from both and implements both abstract methods.

```python
from abc import ABC, abstractmethod

class Printable(ABC):
    @abstractmethod
    def print_document(self):
        pass

class Scannable(ABC):
    @abstractmethod
    def scan_document(self):
        pass

class PrinterScanner(Printable, Scannable):
    def print_document(self):
        print("Printing document")

    def scan_document(self):
        print("Scanning document")

obj = PrinterScanner()
obj.print_document()
obj.scan_document()
```

30. (Problem) Create class A with static method, class B with abstract method, and class C
that inherits from both. Implement the abstract method and call the static method.

```python
from abc import ABC, abstractmethod

class A:
    @staticmethod
    def message():
        print("Static method from A")

class B(ABC):
    @abstractmethod
    def display(self):
        pass

class C(A, B):
    def display(self):
        print("Abstract method implemented in C")

obj = C()

obj.display()
C.message()
```
