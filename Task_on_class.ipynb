```python
#23. Create class Student with private __marks. Add getter and setter methods.
class Student:

    def __init__(self, marks):
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        self.__marks = marks


s1 = Student(80)

print(s1.get_marks())

s1.set_marks(95)

print(s1.get_marks())
```

```python
#24. In class BankAccount, add getter and setter for private __balance with validation.
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):

        if balance >= 0:
            self.__balance = balance
            print("Balance updated")
        else:
            print("Invalid balance")


account = BankAccount(5000)

account.set_balance(7000)

print(account.get_balance())
```

```python
#25. Create class Employee with private __salary. Setter should not allow value below 5000.
class Employee:

    def __init__(self, salary):
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):

        if salary >= 5000:
            self.__salary = salary
            print("Salary updated")
        else:
            print("Salary cannot be below 5000")


e1 = Employee(10000)

e1.set_salary(15000)

print(e1.get_salary())
```

```python
#26. Define class Temperature with private __celsius. Add getter and setter with conversion.
class Temperature:

    def __init__(self, celsius):
        self.__celsius = celsius

    def get_celsius(self):
        return self.__celsius

    def set_celsius(self, value):
        self.__celsius = value

    def get_fahrenheit(self):
        return (self.__celsius * 9 / 5) + 32


t1 = Temperature(25)

print("Celsius:", t1.get_celsius())
print("Fahrenheit:", t1.get_fahrenheit())

t1.set_celsius(30)

print("New Celsius:", t1.get_celsius())
```

```python
#27. Create class Person with private __age. Add setter with validation (age between 0-120).
class Person:

    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):

        if 0 <= age <= 120:
            self.__age = age
            print("Age updated")
        else:
            print("Invalid age")


p1 = Person(25)

p1.set_age(30)

print(p1.get_age())
```

```python
#28. In class Product, add getter and setter for private __price with discount logic.class Product:
class Product:

    def __init__(self, price):
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):

        if price > 0:
            self.__price = price
        else:
            print("Invalid price")

    def apply_discount(self, discount):

        self.__price = self.__price - (
            self.__price * discount / 100
        )


p1 = Product(1000)

p1.apply_discount(10)

print(p1.get_price())
```

```python
#29. Create class Mobile with private __price. Add setter that prints message on price change.
class Mobile:

    def __init__(self, price):
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, new_price):

        if new_price > 0:
            old_price = self.__price
            self.__price = new_price

            print(
                "Price changed from",
                old_price,
                "to",
                self.__price
            )

        else:
            print("Invalid price")


m1 = Mobile(50000)

m1.set_price(45000)
```

```python
#30. Define class Rectangle with private __length and __width. Add getter/setter and area method.
class Rectangle:

    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def get_length(self):
        return self.__length

    def set_length(self, length):

        if length > 0:
            self.__length = length
        else:
            print("Invalid length")

    def get_width(self):
        return self.__width

    def set_width(self, width):

        if width > 0:
            self.__width = width
        else:
            print("Invalid width")

    def area(self):
        return self.__length * self.__width


r1 = Rectangle(10, 5)

print("Area:", r1.area())

r1.set_length(20)

print("New Area:", r1.area())
```
