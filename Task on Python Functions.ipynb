## Here are 5 Hard Questions on Python Functions
1. (Hard) Write a function process_data(*args, **kwargs) that:
• Takes any number of positional arguments (numbers)
• Takes keyword arguments where keys are operation names ("add", "multiply",
"power")
• Performs the requested operations on all numbers
• Returns a dictionary with results Example: process_data(2, 3, 4, add=True,
multiply=True) should return appropriate results.

```python
def process_data(*args, **kwargs):
    result = {}

    if kwargs.get("add") == True:
        result["add"] = sum(args)

    if kwargs.get("multiply") == True:
        mul = 1
        for num in args:
            mul = mul * num
        result["multiply"] = mul

    if kwargs.get("power") == True:
        power_result = []
        for num in args:
            power_result.append(num ** 2)
        result["power"] = power_result

    return result

print(process_data(2, 3, 4, add=True, multiply=True, power=True))
```

 2. (Hard) Create a function make_multiplier(n) that returns another function which
multiplies any number by n. (Use closure concept). Then create double = make_multiplier(2)
and triple = make_multiplier(3) and test them.

```python
def make_multiplier(n):
    def multiplier(x):
        return x * n

    return multiplier

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))
print(triple(5))
```

3. (Hard) Write a function smart_divide(func) as a decorator that:
• Takes a function as argument
• Checks if the second argument is zero before division
• If zero, prints "Cannot divide by zero!" and returns None
• Otherwise, calls the original function

```python
def smart_divide(func):
    def wrapper(a, b):
        if b == 0:
            print("Cannot divide by zero!")
            return None
        else:
            return func(a, b)

    return wrapper


@smart_divide
def divide(a, b):
    return a / b


print(divide(10, 2))
print(divide(10, 0))
```

4. (Hard) Create a function flexible_calc(*args, operation="add", **kwargs) that:
• Can take any number of numbers
• Supports operations: add, subtract, multiply, divide
• If round_result=True in kwargs, round the final answer to 2 decimal places
• Handle all edge cases

```python
def flexible_calc(*args, operation="add", **kwargs):
    if len(args) == 0:
        return "No numbers given"

    if operation == "add":
        result = sum(args)

    elif operation == "subtract":
        result = args[0]
        for num in args[1:]:
            result = result - num

    elif operation == "multiply":
        result = 1
        for num in args:
            result = result * num

    elif operation == "divide":
        result = args[0]
        for num in args[1:]:
            if num == 0:
                return "Cannot divide by zero"
            result = result / num

    else:
        return "Invalid operation"

    if kwargs.get("round_result") == True:
        result = round(result, 2)

    return result


print(flexible_calc(10, 20, 30, operation="add"))
print(flexible_calc(100, 5, 2, operation="divide"))
print(flexible_calc(10, 3, operation="divide", round_result=True))
print(flexible_calc(5, 4, 3, operation="multiply"))
```

5. (Hard) Write a recursive function flatten_list(nested_list) that takes a nested list (list inside
list) and returns a single flat list. Example: flatten_list([1, [2, 3], [4, [5, 6]]]) → [1, 2, 3, 4, 5, 6]

```python
def flatten_list(nested_list):
    flat_list = []

    for item in nested_list:
        if type(item) == list:
            flat_list = flat_list + flatten_list(item)
        else:
            flat_list.append(item)

    return flat_list


print(flatten_list([1, [2, 3], [4, [5, 6]]]))
```
